```python
import cv2 as cv
import matplotlib.pylab as plt
```

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
img = cv.imread("/content/drive/MyDrive/PCVK/KTM.jpg")
img = cv.cvtColor(img, cv.COLOR_BGR2RGB)
img = cv.resize(img, (0, 0), fx=0.5, fy=0.5)

# POLINEMA
cv.rectangle(img, pt1=(100, 20), pt2=(320, 85), color=(0, 0, 0), thickness=2)
# BNI
cv.rectangle(img, pt1=(630, 27), pt2=(770,85), color=(0, 0, 0), thickness=2)
# PRODI
cv.rectangle(img, pt1=(165, 480), pt2=(450, 450), color=(0, 0, 0), thickness=2)

plt.imshow(img)
plt.show()
```
